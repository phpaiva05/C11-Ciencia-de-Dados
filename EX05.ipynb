**Questão 1**

```python
import numpy as np

dataset = np.loadtxt('space.csv', delimiter = ';', dtype=str, encoding='utf-8')
header = dataset[0]
body = dataset[1:]
sucess = body[:, 7] == 'Success'
qtd_sucesso = np.sum(sucess)
qtd_total = len(body)
porcentagem = (qtd_sucesso / qtd_total) * 100

print(f"Porcentagem de sucesso: {porcentagem:.2f}%")
```
