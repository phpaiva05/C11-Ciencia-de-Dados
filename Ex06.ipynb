**Ex1**

```python
import pandas as pd
import numpy as np

seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

print(seriesAno1)
print(seriesAno2)
```

**Ex2**

```python
print(seriesAno1.sum())
print(seriesAno2.sum())
```

**Ex3**

```python
crescimento = seriesAno2 - seriesAno1
print(crescimento)
```

**Ex 4**

```python
print(crescimento[crescimento > 0])
```

**Ex 5**

```python
projecao = seriesAno2 + (crescimento * 2)
print(projecao.nlargest(1))
```

**Ex 6**

```python
df = pd.DataFrame(
    index=['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4])
)

media_x = df['X'][df['X'] < 30].mean()
print(media_x)
```

**Ex 7**

```python
print(df.loc['D'].mean())
print(df.iloc[4].sum())
```

**Ex 8**

```python
slicing = df.loc[['A', 'C', 'E'], ['X', 'Y']]
print(slicing)
print(slicing.sum(axis=1))
print(slicing.sum(axis=0))
```
