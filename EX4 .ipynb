**Ex 2**

```python
mtz = np.zeros([2, 2])
mtz[np.random.randint(0, 2), np.random.randint(0, 2)] = 1

contador_seguros = 0
jogando = True

while jogando:
    linha = int(input("Insira a linha (0 ou 1): "))
    coluna = int(input("Insira a coluna (0 ou 1): "))

    if mtz[linha, coluna] == 1:
        print("Você perdeu.")
        jogando = False
    elif mtz[linha, coluna] == 0:
        print("Posição segura!")
        contador_seguros += 1
        mtz[linha, coluna] = -1
        if contador_seguros == 3:
            print("Parabéns! Você venceu!")
            jogando = False
    else:
        print("Você já escolheu essa posição, tente outra!")

print(mtz)
```

**Ex3**

```python
matriz = np.zeros((3, 5))
linhas, colunas = matriz.shape
total_elementos = linhas * colunas

if total_elementos % 2 == 0:
    classificacao = "par"
else:
    classificacao = "ímpar"

print(matriz)
print(f"Tamanho da matriz: {linhas} linhas e {colunas} colunas.")
print(f"Multiplicação (linhas x colunas): {total_elementos}")
print(f"Esta matriz poderia se tornar um vetor unidimensional com um número {classificacao} de elementos.")
```

**Ex4**

```python
np.random.seed(10)
matri = np.random.randint(1, 51, [4, 4])
medialinha = np.mean(matri, axis=1)
mediacoluna = np.mean(matri, axis=0)
print(matri)
print()
print(medialinha)
print(mediacoluna)

print()
print(max(medialinha))
print(max(mediacoluna))

valores, contagens = np.unique(matri, return_counts=True)
for valor, contagem in zip(valores, contagens):
    print(f"O número {valor} aparece {contagem} vez(es)")

numeros_duplicados = valores[contagens == 2]

print("\nNúmeros que aparecem exatamente 2 vezes:", numeros_duplicados)

```

```python
matriz_zeros = np.zeros([3, 3])
matriz_uns = np.ones([3, 3])
soma = matriz_zeros + matriz_uns
resultado = soma * 5
vetor = resultado.reshape(9)
print("Matriz de Zeros:")
print(matriz_zeros)
print()

print("Matriz de Uns:")
print(matriz_uns)
print()

print("Soma das matrizes:")
print(soma)
print()

print("Resultado multiplicado pelo escalar 5:")
print(resultado)
print()

print("Array unidimensional (reshaped):")
print(vetor)
```

**Ex6**

```python
vetor = np.arange(10, 31, 2)

maior_valor = np.max(vetor)
indice_maior = np.argmax(vetor)
print(maior_valor)
print(indice_maior)

menor_valor = np.min(vetor)
indice_menor = np.argmin(vetor)
print(menor_valor)
print(indice_menor)

soma_total = np.sum(vetor)
print(soma_total)
```
