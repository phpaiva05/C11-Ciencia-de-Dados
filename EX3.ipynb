**Questão 5**

```python
n = int(input("Quantas pessoas você deseja cadastrar? "))

soma_idades = 0
mulheres_menos_20 = 0

for i in range(n):
    print(f"\n--- Dados da {i+1}ª Pessoa ---")
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    sexo = input("Sexo [M/F]: ").strip().upper()

    soma_idades += idade

    if sexo == 'F' and idade < 20:
        mulheres_menos_20 += 1

print("\n" + "="*30)
if n > 0:
    media_idade = soma_idades / n
    print(f"A média de idade do grupo é de {media_idade:.1f} anos.")
    print(f"Quantidade de mulheres com menos de 20 anos: {mulheres_menos_20}")
else:
    print("Nenhuma pessoa foi cadastrada.")
```

**Questão 8**

```python
nomeprod1 = input("Insira o nome do primeiro produto: ")
precoprod1 = float(input("Insira o preço do primeiro produto: "))
quantidadeprod1 = int(input("Insira a quantidade do primeiro produto: "))

nomeprod2 = input("Insira o nome do segundo produto: ")
precoprod2 = float(input("Insira o preço do segundo produto: "))
quantidadeprod2 = int(input("Insira a quantidade do segundo produto: "))

nomeprod3 = input("Insira o nome do terceiro produto: ")
precoprod3 = float(input("Insira o preço do terceiro produto: "))
quantidadeprod3 = int(input("Insira a quantidade do terceiro produto: "))


dictprod = [
    {"nome": nomeprod1, "preco": precoprod1, "quantidade": quantidadeprod1},
    {"nome": nomeprod2, "preco": precoprod2, "quantidade": quantidadeprod2},
    {"nome": nomeprod3, "preco": precoprod3, "quantidade": quantidadeprod3}
]

print("\n--- Resumo do Estoque ---")

for produto in dictprod:
    valor_total = produto["preco"] * produto["quantidade"]
    print(f'Produto: {produto["nome"]} | Valor total em estoque: R$ {valor_total:.2f}')
```
