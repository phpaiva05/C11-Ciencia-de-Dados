**Questão 2**:  Mostre a tabuada de um número que o usuário escolher dentro de um intervalo específico também escolhido por ele.

```python
numero = int(input("Digite o número do qual deseja a tabuada: "))
intervalo_inicio = int(input("Digite o intervalo de inicio da tabuada: "))
intervalo_fim = int(input("Digite o intervalo final da tabuada: "))
for i in range(intervalo_inicio, intervalo_fim + 1):
  print(numero*i)
```

**Questão 3** Faça um programa que leia o sexo de uma pessoa e diga se ela é homem (caso seja
digitado M) ou mulher (caso seja digitado F). Caso seja digitado algo inválido, continue
perguntando até que o usuário entre com um sexo válido.


```python
sexo = input("Entre com seu sexo (F para feminino e M para masculino)")
while sexo != 'F' and sexo != 'M':
  sexo = input("Entre com seu sexo (F para feminino e M para masculino)")
```

**Questão 7**

```python
palavra = input("Digite uma palavra: ")
vogal_count = 0
vogais = "AEIOUaeiou"

for char in palavra:
  print(char.upper())
  if char in vogais:
    vogal_count += 1

print(f"A palavra tem {vogal_count} vogais.")

if 'a' in palavra.lower():
  print("A letra 'a' está presente na palavra.")
else:
  print("A letra 'a' não está presente na palavra.")
```

**Questão 8**

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

adicao = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2
potencia = num1 ** num2
divisao = num1 / num2
resto = num1 % num2

print(adicao)
print(subtracao)
print(multiplicacao)
print(potencia)
print(divisao)
print(resto)

if adicao % 2 == 0:
    print(f"O resultado da adição ({adicao}) é PAR.")
else:
    print(f"O resultado da adição ({adicao}) é ÍMPAR.")
```
